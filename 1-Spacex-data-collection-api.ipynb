# SpaceX Data Collection with API

**Purpose:** document the API-based data collection workflow used in the capstone and produce an analysis-ready launch table.

The original lab uses an HTTP GET request and `pandas.json_normalize`. This repository also includes a local dataset so the notebook remains reproducible when a public API is unavailable.

````python
from pathlib import Path
def resolve_data(filename):
    candidates = [Path("data")/filename, Path("../data")/filename]
    for p in candidates:
        if p.exists():
            return p
    raise FileNotFoundError(filename)

import pandas as pd, json
data_path = resolve_data("spacex_launch_data.csv")
local = pd.read_csv(data_path)
print("Loaded rows:", len(local))
print("Columns:", list(local.columns))
local.head()
````

## API method used in the project

import requests
