# SpaceX Data Collection with API
**Objective:** demonstrate the API collection workflow used in the capstone.

Steps: send a GET request, normalize JSON, select launch attributes, filter to Falcon 9, and save a tabular dataset. The notebook also contains a local fallback so the repository remains reproducible if the public API is unavailable.

```python
import requests, pandas as pd
from pathlib import Path
api_url = "https://api.spacexdata.com/v4/launches/past"
try:
    r = requests.get(api_url, timeout=20)
    r.raise_for_status()
    raw = pd.json_normalize(r.json())
    print("API rows:", len(raw))
except Exception as exc:
    print("API unavailable; using repository dataset:", exc)
    raw = pd.read_csv("../data/spacex_launch_data.csv")
raw.head()
```
