# Predictive Analysis (Classification)
Compare Logistic Regression, SVM, Decision Tree and KNN after encoding and scaling features.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
df=pd.read_csv('../data/sample_launches.csv')
X=pd.get_dummies(df[['LaunchSite','Orbit','PayloadMass']])
y=df['Class']
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.3,random_state=2,stratify=y)
sc=StandardScaler(); Xtr=sc.fit_transform(Xtr); Xte=sc.transform(Xte)
m=DecisionTreeClassifier(random_state=2).fit(Xtr,ytr)
print('Test accuracy:',accuracy_score(yte,m.predict(Xte)))
```
