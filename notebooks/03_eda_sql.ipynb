# EDA with SQL
The capstone uses SQL aggregation to examine launch site, orbit, and landing outcomes.

```python
import sqlite3, pandas as pd
from pathlib import Path
df=pd.read_csv('../data/sample_launches.csv')
con=sqlite3.connect(':memory:')
df.to_sql('SPACEXTBL',con,index=False)
pd.read_sql_query('SELECT LaunchSite, COUNT(*) AS launches FROM SPACEXTBL GROUP BY LaunchSite ORDER BY launches DESC',con)
```
