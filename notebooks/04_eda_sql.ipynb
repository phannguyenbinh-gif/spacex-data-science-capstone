# Exploratory Data Analysis with SQL
This notebook loads the project dataset into SQLite and answers launch-site, orbit and landing-outcome questions with reproducible SQL queries.

```python
import sqlite3, pandas as pd
df = pd.read_csv("../data/spacex_launch_data.csv")
con = sqlite3.connect(":memory:")
df.to_sql("SPACEXTBL", con, index=False, if_exists="replace")
pd.read_sql_query("SELECT LaunchSite, COUNT(*) AS Launches FROM SPACEXTBL GROUP BY LaunchSite ORDER BY Launches DESC", con)
```

```python
pd.read_sql_query("SELECT Orbit, COUNT(*) AS Launches, ROUND(AVG(Class),3) AS SuccessRate FROM SPACEXTBL GROUP BY Orbit ORDER BY Launches DESC", con)
```

```python
pd.read_sql_query("""SELECT LaunchSite,
COUNT(*) AS Total,
SUM(CASE WHEN Class=1 THEN 1 ELSE 0 END) AS Successful,
ROUND(AVG(Class)*100,1) AS SuccessPct
FROM SPACEXTBL GROUP BY LaunchSite ORDER BY SuccessPct DESC""", con)
```

## Interpretation
SQL aggregation makes the EDA auditable: site frequency, orbit frequency and historical landing success can be reproduced directly from the underlying table rather than inferred from a chart alone.
