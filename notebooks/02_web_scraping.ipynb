# Web Scraping Falcon 9 Launch Data

**Purpose:** demonstrate the BeautifulSoup parsing workflow used to extract launch records from HTML tables.

For reproducibility, this notebook executes against a local Falcon 9 sample table while preserving the same parsing logic used for a web page.

````python
from pathlib import Path
def resolve_data(filename):
    candidates = [Path("data")/filename, Path("../data")/filename]
    for p in candidates:
        if p.exists():
            return p
    raise FileNotFoundError(filename)

from bs4 import BeautifulSoup
import pandas as pd
html = resolve_data("falcon9_sample.html").read_text(encoding="utf-8")
soup = BeautifulSoup(html, "html.parser")
print("Page title:", soup.title.get_text(strip=True))
````

````python

records=[]
for tr in soup.select("table.wikitable tr"):
    cells=[c.get_text(" ", strip=True) for c in tr.find_all(["th","td"])]
    if cells:
        records.append(cells)

headers=records[0]
scraped=pd.DataFrame(records[1:], columns=headers)
print("Scraped rows:", len(scraped))
scraped
````
