# Folium Launch-Site Geographic Analysis

**Purpose:** create an interactive map of representative Falcon 9 launch facilities and interpret their geographic context.

```python

import folium
from pathlib import Path

sites=[
    ("CCAFS SLC 40",28.563197,-80.576820),
    ("KSC LC 39A",28.573255,-80.646895),
    ("VAFB SLC 4E",34.632834,-120.610745)
]
m=folium.Map(location=[29.5,-95], zoom_start=4)
for name,lat,lon in sites:
    folium.Marker([lat,lon], popup=name, tooltip=name).add_to(m)
    folium.Circle([lat,lon], radius=1000, color="blue", fill=False).add_to(m)
m
```

```python
from pathlib import Path
def resolve_data(filename):
    candidates = [Path("data")/filename, Path("../data")/filename]
    for p in candidates:
        if p.exists():
            return p
    raise FileNotFoundError(filename)

map_path = resolve_data("spacex_launch_data.csv").parent / "falcon9_launch_sites_map.html"
m.save(str(map_path))
print("Saved map:", map_path)
```

## Geographic insight

The mapped launch facilities are coastal. Folium adds operational/geographic context that is not visible in tabular analysis and supports comparison among launch locations.
