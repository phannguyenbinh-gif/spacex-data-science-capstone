# Interactive Visual Analytics with Folium
**Objective:** map launch facilities and add geospatial context.

```python
import folium
sites = [
    ("CCAFS SLC 40", 28.563197, -80.576820),
    ("KSC LC 39A", 28.573255, -80.646895),
    ("VAFB SLC 4E", 34.632834, -120.610745),
]
m = folium.Map(location=[29.5,-95], zoom_start=4)
for name,lat,lon in sites:
    folium.Circle([lat,lon], radius=1000, color="blue", fill=True,
                  popup=name, tooltip=name).add_to(m)
m
```

## Interpretation
The launch sites are coastal. Mapping provides context that tabular EDA cannot: launch facilities can be interpreted relative to coastline, transport routes and nearby infrastructure. The map is interactive in Jupyter.

```python
m.save("../data/falcon9_launch_sites_map.html")
print("Saved interactive map to data/falcon9_launch_sites_map.html")
```
