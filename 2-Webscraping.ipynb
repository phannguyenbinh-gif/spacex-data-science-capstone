# Data Collection with Web Scraping
**Objective:** show the BeautifulSoup workflow for extracting Falcon 9 launch-history tables.

The production workflow requests the launch-history page, parses HTML, identifies table rows/cells and converts records into a Pandas DataFrame.

```python
import requests
from bs4 import BeautifulSoup
url = "https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches"
html = requests.get(url, timeout=20).text
soup = BeautifulSoup(html, "html.parser")
print(soup.title)
tables = soup.find_all("table")
print("HTML tables found:", len(tables))
```

## Parsing logic
For each launch table, iterate through `tr` rows and read `th`/`td` cells. Clean whitespace, standardize dates/numeric values, and store the result in a DataFrame. The parsed data can then be joined with API-derived launch attributes.

```python
rows = []
for tr in soup.select("table.wikitable tr")[:8]:
    cells = [c.get_text(" ", strip=True) for c in tr.find_all(["th","td"])]
    if cells:
        rows.append(cells)
print("Example parsed rows:", len(rows))
rows[:2]
```
